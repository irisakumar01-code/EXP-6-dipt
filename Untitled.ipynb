EDGE DETECTION USING OPENCV

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('city.jpg')

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original City Image")
plt.axis("off")
plt.show()
```

Step 3: Convert Image to Grayscale

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
sobel = cv2.Sobel(gray, cv2.CV_8U, 1, 1)

plt.imshow(sobel, cmap='gray')
plt.title("Sobel")
plt.axis('off')
plt.show()
```

Step 5: Apply Prewitt Edge Detector

```python
kernel = np.array([[-1, 0, 1],
                   [-1, 0, 1],
                   [-1, 0, 1]])

prewitt = cv2.filter2D(gray, -1, kernel)

plt.imshow(prewitt, cmap='gray')
plt.title("Prewitt")
plt.axis('off')
plt.show()
```

Step 6: Apply Roberts Edge Detector

```python
kernel = np.array([[1, 0],
                   [0, -1]])

roberts = cv2.filter2D(gray, -1, kernel)

plt.imshow(roberts, cmap='gray')
plt.title("Roberts")
plt.axis('off')
plt.show()
```

Step 7: Laplacian Edge Detection

```python
laplacian = cv2.Laplacian(gray, cv2.CV_8U)

plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian")
plt.axis('off')
plt.show()
```

Step 8: Canny Edge Detection

```python
canny = cv2.Canny(gray, 100, 200)

plt.imshow(canny, cmap='gray')
plt.title("Canny")
plt.axis('off')
plt.show()
```

Step 9: Display All Results Together

```python
images = [sobel, prewitt, roberts, laplacian, canny]
titles = ["Sobel", "Prewitt", "Roberts", "Laplacian", "Canny"]

plt.figure(figsize=(12, 5))

for i in range(5):
    plt.subplot(1, 5, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.show()
```
